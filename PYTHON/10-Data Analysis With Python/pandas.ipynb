### Pandas-dataFrame And Series

Pandas is a powerful data manipulation library in python, widely used for data analysis and data cleaning. it provides two primary data structures: Series and dataFrame . A Series is a one-dimentional array-like object, while a DataFrame is a two-Dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

```python
import pandas as pd
```

```python
## Series
##A Pandas Series is a one-demensional array-like object that can hold any data type. It is similar to a column in a table.

import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print("series \n",series)
```

```python
## Create a Series from dictionary
data={'a':1,'b':2,'c':3}
series_data=pd.Series(data)
print(series_data)
```

```python
data=[10,20,30]
index=['a','b','c']
pd.Series(data,index=index)
```

```python
## dataframe
## create a Dataframe from a dictionary oof list

data={
      'Name':['Krish','John','Jack'],
      'Age':[25,30,45],
      'City':['Bangaloore','New York','Florida']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
```

```python
## Create a Data frame From a List of Dictionaries

data=[
    {'Name':'Krish' ,'Age':32,'City':'Bangalore'},
    {'Name':'John'  ,'Age':34,'City':'Bangalore'},
    {'Name':'vanie' ,'Age':38,'City':'Bangalore'},
    {'Name':'bappie','Age':31,'City':'Bangalore'}, 
]
df=pd.DataFrame(data)
print(df)
print(type(df))
```
