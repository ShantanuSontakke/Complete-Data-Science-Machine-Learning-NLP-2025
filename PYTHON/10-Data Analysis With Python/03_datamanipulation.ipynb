### Data Manipulation and Analysis with Pandas
 

Data manipulation and analysis  are key tasks in any data science or data analysis project. pandas provides a wide range of functions for data manipulation and analysis making it easier to clean, tranform and extract insights from data. in this lesson, we will cover various data manipulation and analysis techniques using pandas.

```python
import pandas as pd
```

```python
## handling missing values
df.isnull()
```

```python
df.isnull().any()
```

```python
# misssing values sum
df.isnull().sum()
```

```python
df_filled=df.fillna(0)
```

```python
## filling missing values with the mean of the column
df['Sales_fillNA']=df['sales'].fillna(df['sales'].mean())
df
```

```python
df.dtypes
```

```python
### renaming columns
df=df.rename(columns={'date':'Sales Date'})
df
```

```python
## change datatypes
df['Value_new']=df['value'].fillna(df['value'].mean()).astype(int)
df.head()
```

```python
df['New Value']=df['value'].apply(lambda x:x**2)
df.head()
```

```python
grouped_mean=df.groupby('product')['value'].mean()
print(grouped_mean)
```

```python
grouped_sum=df.groupby(['product','region'])['value'].sum()
print(grouped_sum)
```

```python
## aggregate multiple functions
grouped_agg=df.groupby('region')['value'].agg(['mean','sum','count'])
grouped_agg
```

```python
### merging and joining Dataframes
# Create sample Dataframe
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'],'Value1': [1,2,3,]})
df2 = pd.DataFrame({'Key': ['A','B','D'],'value2': [4,5,6]})
```

```python
df1
```

```python
df2
```

```python
# Merge Dataframe on the 'Key value'
pd.merge(df1,df2,on="Key",how="inner")  #inner means same parameter like A B are both in key df1 and df2
```

```python
pd.merge(df1,df2,on="Key",how="outer") 
```

```python
pd.merge(df1,df2,on="Key",how="left") 
```

```python
pd.merge(df1,df2,on="Key",how="right") 
```
