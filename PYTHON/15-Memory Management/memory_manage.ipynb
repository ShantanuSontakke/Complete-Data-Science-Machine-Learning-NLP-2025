### Reference Counting

Reference counting is the primary method Python uses to manage memory.Each object in Python maintains a count of reference pointing to it.When the reference count drops to zero ,the memory occupied by the object is deallocated.

```python
import sys

a = []
# 2 (one reference from a and one from getrefcount ())
print(sys.getrefcount(a))
```

```python
b=a
print(sys.getrefcount(b))  # 3 references now
```

```python
del b
print(sys.getrefcount(a))  # back to 2 references
```

#### Garbage Collection

Python includes a cyclic garbage collector to handler refernce cycles. Refernce cycles occur when objects refernce each other, preventing their reference counts from reaching zero.

```python
import gc
## enable garbage collection
gc.enable()

```

```python
gc.disable()
```

```python
gc.collect()
```

```python
### Get garbage collector stats
print(gc.get_stats())
```

```python
### get unreachable objects
print(gc.garbage)
```

### Memory Mangement Best Practices
1.Use Local Variabls: Local variable have a shorter lifespan and are freed sooner than global variables.
2.Avoid Circular Refernces: Circular refernce can lead to memory leaks if not properly managed.
3.Use generators:Generators produce items one at a time and only keep one item in memory at a time,making them memory efficient.
4.Explicity Delete objects:Use the del statement to delete variables and objects explicity.
5.Profile Memory usage: Use memory profiling tools like tracemalloc and memory_profiler to identify memory leaks and optimize memory usage.

```python
import gc

class MyObject:
    def __init__(self, name):
        self.name = name
        print(f"Object {self.name} created")

    def __del__(self):
        print(f"Object {self.name} deleted")

## create circular refernces
obj1 = MyObject("obj1")
obj2 = MyObject("obj2")
obj1.ref = obj2
obj2.ref = obj1


del obj1
del obj2


## Manually trigger the garbage collection
gc.collect()
```

```python
# Print collected objects
print(f"Garbage collected: {gc.garbage}")
```

```python
### Generators for memory Efficiency
#Generators allow you to produce item one at a time, using memory efficiency by only kepping one item in memory at a time.

def generate_numbers(n):
    for i in range(n):
        yield i

## using the generator
for num in generate_numbers(100000):
    print(num)
    if num>10:
        break
```

```python
## profiling Memory Usage with tracemalloc
import tracemalloc

def create_list():
    return [i for i in range(10000)]

def main():
    tracemalloc.start()

    create_list()

    snapshot = tracemalloc.take_snapshot()
    top_stats = snapshot.statistics('lineno')

    print("[ Top 10 ]")
    for stat in top_stats[:10]:
        print(stat)
```

```python
main()
```
